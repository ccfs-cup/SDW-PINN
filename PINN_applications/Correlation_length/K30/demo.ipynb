```python
import torch
import matplotlib.pyplot as plt
import numpy as np

# Create the x coordinates
x = torch.linspace(0, 100, 200)  # x coordinates

# Define fixed y values for line plots along y=30, y=50, and y=70
y1 = 30
y2 = 50
y3 = 70

# Calculate the Gaussian source values
source_term1 = 500 / (2 * torch.pi * 0.5**2) * torch.exp(-((x - 30)**2 + (y1 - 30)**2) / (2 * 0.5**2))
source_term2 = 600 / (2 * torch.pi * 2**2) * torch.exp(-((x - 30)**2 + (y2 - 70)**2) / (2 * 2**2))
source_term3 = 700 / (2 * torch.pi * 2**2) * torch.exp(-((x - 70)**2 + (y3 - 50)**2) / (2 * 2**2))

# Create the line plots
plt.figure(figsize=(8, 6))

# Plot source_term1
plt.plot(x.numpy(), source_term1.numpy(), label='Source 1', color='r')

# Plot source_term2
plt.plot(x.numpy(), source_term2.numpy(), label='Source 2', color='g')

# Plot source_term3
plt.plot(x.numpy(), source_term3.numpy(), label='Source 3', color='b')

# Add the title, axis labels, and legend
plt.title('Gaussian Sources as Line Plots')
plt.xlabel('X Coordinate')
plt.ylabel('Source Strength')
plt.legend()

# Display the figure
plt.grid(True)
plt.show()
```
